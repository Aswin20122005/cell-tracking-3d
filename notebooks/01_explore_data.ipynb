# 01 — Explore the data

Quick look at the 3D volumes, cell-centre labels, and a max-projection.
Uses synthetic data if the real dataset isn't present.

```python
import sys, os
sys.path.append(os.path.abspath('..'))
import numpy as np
import matplotlib.pyplot as plt
from src.data.dataset import EmbryoDataset
```

```python
ds = EmbryoDataset(root='../data', embryos=['embryo_01'], synthetic=True)
frames = ds.frames()
print(f'{len(frames)} frames')
f = frames[0]
print('volume shape:', f.volume.shape, '| cells:', len(f.centres))
```

```python
# Max-intensity projection along Z with cell centres overlaid
mip = f.volume.max(axis=0)
plt.figure(figsize=(6, 6))
plt.imshow(mip, cmap='magma')
plt.scatter(f.centres[:, 2], f.centres[:, 1], s=20, edgecolor='cyan', facecolor='none')
plt.title('Z max-projection + cell centres')
plt.axis('off')
plt.show()
```
